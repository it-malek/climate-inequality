```python
import plotly.express as px
import pandas as pd
import geopandas as gpd
```

```python
df = pd.read_csv('GlobalLandTemperaturesByCity.csv')
```

```python
df['Year'] = pd.DatetimeIndex(df['dt']).year
```

```python
def proper_l(x):
    if (x.endswith("N")) or (x.endswith("E")):
        return (x[0:-1])
    else:
        return ("-" + x[0:-1])
df["Latitude"] = df["Latitude"].apply(proper_l)
df["Longitude"] = df["Longitude"].apply(proper_l)
df['Longitude'] = df['Longitude'].astype(float, errors = 'raise')
df['Latitude'] = df['Latitude'].astype(float, errors = 'raise')
df.info()
df
```

```python
df.to_csv('GlobalLandTemperatureByCity1.csv')
```
